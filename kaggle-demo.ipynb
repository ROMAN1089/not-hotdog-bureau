```python
from transformers import pipeline
from PIL import Image
from pathlib import Path
detector = pipeline('image-classification', model='julien-c/hotdog-not-hotdog', device=-1)
images = list(Path('/kaggle/input').rglob('*.jpg')) + list(Path('/kaggle/input').rglob('*.png'))
if not images:
    print('Добавьте датасет с фото через Add Input и запустите ячейку снова.')
else:
    image = Image.open(images[0]).convert('RGB')
    display(image)
    print(detector(image, top_k=2))
```
